```python
import serpapi

client = serpapi.Client(api_key="f82c5216560ea23b9f7dfdfdb5c75af7e6a3b903c8e3daae0155a9f0c0c55e31")
results = client.search({
  "engine": "google_trends",
  "q": "coffee,milk,bread,pasta,steak",
  "data_type": "TIMESERIES"
})
interest_over_time = results["interest_over_time"]

print(interest_over_time)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
# timeline_data (5 day as a single row)
# y will be value of values.value for query q

timeline_data = interest_over_time["timeline_data"]
len(timeline_data)
```

```python
# take 53 as single da
X = [x for x in range(1, 54)]
X_ORIGINAL = X
y = [t["values"][0]["value"] for t in timeline_data]

X = np.array(X)
X = X.reshape(-1,1)

y = np.array(y)

model.fit(X, y)

print(X)
print(y)
```

```python
slope = model.coef_
bias = model.intercept_
```

```python
n = len(y) - 1
y = y.astype(int)

# Growth
growth_last =( y[n] - y[n-1] )/ y[n - 1]
growth_previous_last =( y[n - 2] - y[n - 3] )/ y[n - 3]

# Acceleration
acceleration = False
if (growth_last > growth_previous_last):
  print("Growing")
  acceleration = True
else:
  acceleration = False
  print("NG")

moving_average = (y[n-1] + y[n-2] + y[n-3]) / 3
print("Current trend", moving_average)

peak_distance = 1 - (y[n-1] / 100)
print("Peak Distance: ", peak_distance)
```

```python
future_days = 5
X_future = np.array([x for x in range(len(X_ORIGINAL) + 1, len(X_ORIGINAL) + future_days + 1)]).reshape(-1, 1)
predictions = model.predict(X_future)

print(f"Predictions for the next {future_days} days:")
for i, pred in enumerate(predictions):
    print(f"Day {len(X_ORIGINAL) + i + 1}: {pred:.2f}")
```
